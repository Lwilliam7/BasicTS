# Router Model Results

This notebook displays the router test results saved under `results/best_*` and the combined summary file `results/router_model_results_summary.csv`.

```python
from pathlib import Path
import pandas as pd

try:
    from IPython.display import display
except Exception:
    def display(value):
        print(value)

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

summary_path = ROOT / "results" / "router_model_results_summary.csv"
router_results_df = pd.read_csv(summary_path)

print(f"Loaded {len(router_results_df)} result rows from {summary_path}")
display(router_results_df)
```

## Best Method Per Model Count

```python
best_by_group_df = (
    router_results_df.sort_values(["model_count", "test_mae"])
    .groupby("model_count", as_index=False)
    .first()
)
display(best_by_group_df[["model_count", "selected_models", "method", "test_mae", "test_mse", "test_rmse"]])
```

## Learned Router Only

```python
learned_router_df = router_results_df[
    router_results_df["method"].eq("Learned prediction-aware router")
].sort_values("test_mae")
display(learned_router_df[["model_count", "selected_models", "test_mae", "test_mse", "test_rmse"]])
```

## Overall Ranking

```python
overall_ranking_df = router_results_df.sort_values("test_mae", ignore_index=True)
display(overall_ranking_df[["model_count", "selected_models", "method", "test_mae", "test_mse", "test_rmse"]])
```
