```python
# all in one block
import os
from pathlib import Path

ROOT = Path(r"C:\Users\luwil\OneDrive\Documents\Code\BasicTS") # start working from 
os.chdir(ROOT) # change folders
print("CWD:", Path.cwd()) # write to folder

from basicts.models.DLinear import DLinear, DLinearConfig
from basicts.configs import BasicTSForecastingConfig
from basicts import BasicTSLauncher

inputLen = 96
model_config = DLinearConfig(input_len= inputLen, output_len=inputLen) # use past 336 time steps to predict the next 336 time steps

cfg = BasicTSForecastingConfig( # expieremetn setup
    model=DLinear, 
    model_config=model_config,
    dataset_name="ETTh1",
    gpus="0",
    num_epochs=1,
    ckpt_save_dir="checkpoints/DLinear/my_first_clean_run_96",
)
cfg.dataset_params.input_len = inputLen
cfg.dataset_params.output_len = inputLen
BasicTSLauncher.launch_training(cfg) # launches 
```

```python
BasicTSLauncher.launch_evaluation(
    cfg,
    "checkpoints/DLinear/my_first_clean_run/9426e50751d8a47e46a1f030a608199e/DLinear_best_val_MAE.pt"
)
```
