```python
import json
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 1. Settings
# -----------------------------

ROOT = Path(r"C:\Users\luwil\OneDrive\Documents\Code\BasicTS")
CHECKPOINTS = ROOT / "checkpoints"

print("Checkpoint folder:", CHECKPOINTS)
print("Exists:", CHECKPOINTS.exists())

# -----------------------------
# 2. Find metrics JSON files
# -----------------------------

metric_files = list(CHECKPOINTS.rglob("*metrics*.json"))

print("Found metrics files:", len(metric_files))

# -----------------------------
# 3. Helper functions
# -----------------------------

def get_run_name(file_path):
    parts = file_path.parts
    if "checkpoints" in parts:
        idx = parts.index("checkpoints")
        if idx + 1 < len(parts):
            return parts[idx + 1]
    return "unknown"


def add_row(rows, run_name, model_name, metrics, file_path):
    if not isinstance(metrics, dict):
        return

    mae = metrics.get("MAE")
    mse = metrics.get("MSE")
    rmse = metrics.get("RMSE")
    mape = metrics.get("MAPE")
    wape = metrics.get("WAPE")

    if mae is None and "metrics" in metrics and isinstance(metrics["metrics"], dict):
        inner = metrics["metrics"]
        mae = inner.get("MAE")
        mse = inner.get("MSE")
        rmse = inner.get("RMSE")
        mape = inner.get("MAPE")
        wape = inner.get("WAPE")

    if mae is not None and mse is not None:
        rows.append({
            "run": run_name,
            "model": str(model_name),
            "MAE": float(mae),
            "MSE": float(mse),
            "RMSE": float(rmse) if rmse is not None else None,
            "MAPE": float(mape) if mape is not None else None,
            "WAPE": float(wape) if wape is not None else None,
            "path": str(file_path),
        })


# -----------------------------
# 4. Load all metrics
# -----------------------------

rows = []

for file in metric_files:
    run_name = get_run_name(file)

    try:
        with open(file, "r", encoding="utf-8") as f:
            data = json.load(f)

        # Case 1:
        # {
        #   "Model A": {"MAE": ..., "MSE": ...},
        #   "Model B": {"MAE": ..., "MSE": ...}
        # }
        if isinstance(data, dict):
            has_direct_metrics = "MAE" in data and "MSE" in data

            if has_direct_metrics:
                model_name = (
                    data.get("model")
                    or data.get("model_name")
                    or data.get("name")
                    or file.parent.name
                )
                add_row(rows, run_name, model_name, data, file)

            else:
                for key, value in data.items():
                    if isinstance(value, dict):
                        add_row(rows, run_name, key, value, file)

        # Case 2:
        # [
        #   {"model": "Model A", "MAE": ..., "MSE": ...},
        #   {"model": "Model B", "MAE": ..., "MSE": ...}
        # ]
        elif isinstance(data, list):
            for item in data:
                if isinstance(item, dict):
                    model_name = (
                        item.get("model")
                        or item.get("model_name")
                        or item.get("name")
                        or file.parent.name
                    )
                    add_row(rows, run_name, model_name, item, file)

    except Exception as e:
        print("Could not read:", file)
        print("Error:", e)

# -----------------------------
# 5. Build table
# -----------------------------

df = pd.DataFrame(rows)

if df.empty:
    print("No metrics found. Check that your JSON files contain MAE and MSE.")
else:
    df = df.sort_values("MAE").reset_index(drop=True)

    print("\n=== ALL RESULTS SORTED BY MAE ===")
    display(df)

    # -----------------------------
    # 6. Filter to fair full 12-step results
    # -----------------------------

    full_12_mask = df["model"].str.contains(
        "1-12|full|hybrid|stacking|selector|soft|ridge|gating|blend",
        case=False,
        na=False,
    )

    split_only_mask = df["model"].str.contains(
        "steps 1-6|steps 7-12",
        case=False,
        na=False,
    )

    full_12_df = df[full_12_mask & ~split_only_mask].copy()
    full_12_df = full_12_df.sort_values("MAE").reset_index(drop=True)

    print("\n=== FAIR FULL 12-STEP RESULTS SORTED BY MAE ===")
    display(full_12_df)

    # -----------------------------
    # 7. Best models
    # -----------------------------

    print("\n=== BEST OVERALL RESULT ===")
    display(df.iloc[0])

    if not full_12_df.empty:
        print("\n=== BEST FULL 12-STEP MAE RESULT ===")
        display(full_12_df.loc[full_12_df["MAE"].idxmin()])

        print("\n=== BEST FULL 12-STEP MSE RESULT ===")
        display(full_12_df.loc[full_12_df["MSE"].idxmin()])

    # -----------------------------
    # 8. Plot top 10 full 12-step models by MAE
    # -----------------------------

    if not full_12_df.empty:
        top10 = full_12_df.head(10)

        plt.figure(figsize=(12, 6))
        plt.barh(top10["model"], top10["MAE"])
        plt.xlabel("MAE")
        plt.ylabel("Model")
        plt.title("Top 10 Full 12-Step Models by MAE")
        plt.gca().invert_yaxis()
        plt.tight_layout()
        plt.show()

        top10_mse = full_12_df.sort_values("MSE").head(10)

        plt.figure(figsize=(12, 6))
        plt.barh(top10_mse["model"], top10_mse["MSE"])
        plt.xlabel("MSE")
        plt.ylabel("Model")
        plt.title("Top 10 Full 12-Step Models by MSE")
        plt.gca().invert_yaxis()
        plt.tight_layout()
        plt.show()
```
