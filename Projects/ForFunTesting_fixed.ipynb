```python
# Step 1: Configure notebook environment and repository path.
from pathlib import Path
import os
import sys

repo_root = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "src" / "basicts").exists()
)
sys.path.insert(0, str(repo_root / "src"))
os.chdir(repo_root)

print("Python executable:", sys.executable)
print("Repo root:", repo_root)
```

```python
# Step 2: Import BasicTS modules and verify iTransformer access.
from basicts.configs import BasicTSForecastingConfig
from basicts.launcher import BasicTSLauncher
from basicts.models.iTransformer import iTransformerConfig, iTransformerForForecasting

print("Imported iTransformerForForecasting:", iTransformerForForecasting)
print("Imported iTransformerConfig:", iTransformerConfig)
```

```python
# Step 3: Configure and launch training for iTransformer on the ETTh1 dataset.
input_len = 336
output_len = 336

model_config = iTransformerConfig(
    input_len=input_len,
    output_len=output_len,
    num_features=7,
    hidden_size=32,
    intermediate_size=32,
    n_heads=1,
    num_layers=1,
    dropout=0.1,
    use_revin=True,
)

cfg = BasicTSForecastingConfig(
    model=iTransformerForForecasting,
    model_config=model_config,
    dataset_name="ETTh1",
    input_len=input_len,
    output_len=output_len,
    batch_size=32,
    lr=1e-3,
    num_epochs=1,
    gpus=None,  # Set "0" only if CUDA is available and configured in this kernel.
    ckpt_save_dir="checkpoints/iTransformer/ETTh1_notebook_demo",
    eval_after_train=True,
    save_results=False,
)

BasicTSLauncher.launch_training(cfg)
```
